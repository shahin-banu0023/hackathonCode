```python
import pandas as pd

# Load the dataset
df = pd.read_csv('final_food_delivery_dataset.csv')

# Preprocessing: Convert order_date to datetime
df['order_date'] = pd.to_datetime(df['order_date'], format='%d-%m-%Y')

# --- Q1: Highest total revenue from Gold members by city ---
gold_df = df[df['membership'] == 'Gold']
q1_result = gold_df.groupby('city')['total_amount'].sum().idxmax()
print(f"Q1 (Highest Revenue - Gold): {q1_result}")

# --- Q2: Cuisine with highest average order value ---
q2_result = df.groupby('cuisine')['total_amount'].mean().idxmax()
print(f"Q2 (Highest Avg Order Value Cuisine): {q2_result}")

# --- Q3: Distinct users with total orders > ₹1000 ---
user_totals = df.groupby('user_id')['total_amount'].sum()
q3_count = (user_totals > 1000).sum()
print(f"Q3 (Users with > ₹1000 total): {q3_count}")

# --- Q4: Rating range with highest total revenue ---
bins = [3.0, 3.5, 4.0, 4.5, 5.0]
labels = ['3.0 – 3.5', '3.6 – 4.0', '4.1 – 4.5', '4.6 – 5.0']
df['rating_range'] = pd.cut(df['rating'], bins=bins, labels=labels, include_lowest=True)
q4_result = df.groupby('rating_range', observed=False)['total_amount'].sum().idxmax()
print(f"Q4 (Highest Revenue Rating Range): {q4_result}")

# --- Q5: City with highest average order value among Gold members ---
q5_result = gold_df.groupby('city')['total_amount'].mean().idxmax()
print(f"Q5 (Highest Avg Order Value - Gold City): {q5_result}")

# --- Q6: Cuisine with lowest restaurants but significant revenue ---
cuisine_stats = df.groupby('cuisine').agg(
    distinct_restaurants=('restaurant_id', 'nunique'),
    total_revenue=('total_amount', 'sum')
)
q6_result = cuisine_stats['distinct_restaurants'].idxmin()
print(f"Q6 (Lowest restaurants cuisine): {q6_result}")

# --- Q7: Percentage of total orders placed by Gold members ---
gold_percentage = (len(gold_df) / len(df)) * 100
print(f"Q7 (Gold Order %): {round(gold_percentage)}%")

# --- Q8: Specific restaurant with highest avg order value and < 20 orders ---
q8_options = [
    'Grand Cafe Punjabi', 'Grand Restaurant South Indian',
    'Ruchi Mess Multicuisine', 'Ruchi Foods Chinese'
]
rest_stats = df[df['restaurant_name_order'].isin(q8_options)].groupby('restaurant_name_order').agg(
    avg_order_value=('total_amount', 'mean'),
    order_count=('order_id', 'count')
)
q8_result = rest_stats[rest_stats['order_count'] < 20]['avg_order_value'].idxmax()
print(f"Q8 (Highest Avg Value < 20 orders): {q8_result}")

# --- Q9: Best revenue combination among options ---
q9_options = [
    ('Gold', 'Indian'), ('Gold', 'Italian'),
    ('Regular', 'Indian'), ('Regular', 'Chinese')
]
comb_revenue = df.groupby(['membership', 'cuisine'])['total_amount'].sum()
# Filter for specific options provided in the prompt
q9_scores = {opt: comb_revenue.get(opt, 0) for opt in q9_options}
q9_result = max(q9_scores, key=q9_scores.get)
print(f"Q9 (Highest Revenue Combination): {q9_result}")

# --- Q10: Quarter with highest total revenue ---
df['quarter'] = df['order_date'].dt.quarter
q10_result = df.groupby('quarter')['total_amount'].sum().idxmax()
print(f"Q10 (Highest Revenue Quarter): Q{q10_result}")
```

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('final_food_delivery_dataset.csv')

# 1. Total orders placed by users with Gold membership
gold_orders_count = df[df['membership'] == 'Gold'].shape[0]

# 2. Total revenue (rounded to nearest integer) generated from orders placed in Hyderabad city
hyderabad_revenue = round(df[df['city'] == 'Hyderabad']['total_amount'].sum())

# 3. How many distinct users placed at least one order
distinct_users_count = df['user_id'].nunique()

# 4. Average order value (rounded to 2 decimals) for Gold members
gold_avg_order_value = round(df[df['membership'] == 'Gold']['total_amount'].mean(), 2)

# 5. How many orders were placed for restaurants with rating ≥ 4.5
high_rating_orders_count = df[df['rating'] >= 4.5].shape[0]

# 6. How many orders were placed in the top revenue city among Gold members only?
# Step A: Filter for Gold members
gold_df = df[df['membership'] == 'Gold']
# Step B: Find the city with the highest total revenue among Gold members
top_revenue_city_gold = gold_df.groupby('city')['total_amount'].sum().idxmax()
# Step C: Count orders in that city for Gold members
orders_in_top_gold_city = gold_df[gold_df['city'] == top_revenue_city_gold].shape[0]

# Output Results
print(f"Total orders by Gold members: {gold_orders_count}")
print(f"Total revenue in Hyderabad: ₹{hyderabad_revenue}")
print(f"Distinct users: {distinct_users_count}")
print(f"Average order value (Gold): ₹{gold_avg_order_value}")
print(f"Orders with rating >= 4.5: {high_rating_orders_count}")
print(f"Orders in the top Gold city ({top_revenue_city_gold}): {orders_in_top_gold_city}")
```
