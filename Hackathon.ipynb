```python
import pandas as pd
import sqlite3
```

```python
orders = pd.read_csv("C:/Users/Shahin Banu/Downloads/orders.csv")
orders.head()
```

```python
users = pd.read_json("C:/Users/Shahin Banu/Downloads/users.json")
users.head()
```

```python
conn = sqlite3.connect("restaurants.db")

# Run SQL file once to create table
with open("C:/Users/Shahin Banu/Downloads/restaurants.sql", "r") as f:
    sql_script = f.read()

conn.executescript(sql_script)

restaurants = pd.read_sql("SELECT * FROM restaurants", conn)

restaurants.head()
```

```python
orders_users = orders.merge(
    users,
    on="user_id",
    how="left"
)
```

```python
final_dataset = orders_users.merge(
    restaurants,
    on="restaurant_id",
    how="left"
)
```

```python
final_dataset.to_csv(
    "final_food_delivery_dataset.csv",
    index=False
)
```
